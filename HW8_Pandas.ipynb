```python
# Рахунки на вечерю з сумою чеку більше 30 доларів

expensive_dinners = df[(df['time'] == 'Dinner') & (df['total_bill'] > 30)]
print(len(expensive_dinners))

display(expensive_dinners.head(5))
```

**4. Описова статистика**

Обчисліть (кожен пункт оцінюється в 0.5 бала):
- Середній розмір рахунку на всьому наборі даних
- Максимальні чайові
- Кількість пропущених значень у всіх колонках
- Скільки в наборі даних рахунків за обіди і за вечері

```python
# Середній розмір рахунку на всьому наборі даних
avg = df['total_bill'].mean()
print(f'Average bill - {avg}')
print()

# Максимальні чайові
max_tips = df['tip'].max()
print(f'Max tips - {max_tips}')
print()

# Скільки в наборі даних рахунків за обіди і за вечері
print(df['time'].value_counts())
print()

# Кількість пропущених значень у всіх колонках
df.isnull().sum()
```

**5. Статистика із групуванням**

Обчисліть (кожен пункт оцінюється в 0.5 бала):
- Середній рахунок за статтю (groupby `sex`)
- Середні чайові за днем тижня (groupby `day`)
- Скільки відвідувачів у кожен день

```python
# Середній рахунок за статтю
by_sex = df.groupby('sex')['total_bill'].mean()
print("Average bill by sex:")
by_sex
```

```python
# Середні чайові за днем тижня
by_day = df.groupby('day')['tip'].mean()
print("Average tips by day:")
by_day
```

```python
# Скільки відвідувачів у кожен день
visitors = df.groupby('day')['size'].sum()
print("Total visitors by day:")
visitors
```

Відсортуйте датасет і виведіть результат сортування в кожному випадку окремо (кожен пункт оцінюється в 0.5 бала):

```python
# За розміром рахунку (зростання)
df_sorted_bill = df.sort_values('total_bill', ascending=True)
display(df_sorted_bill.head(5))
```

```python
# За чайовими (спадання)
df_sorted_tips = df.sort_values('tip', ascending=False)
display(df_sorted_tips.head(5))
```
